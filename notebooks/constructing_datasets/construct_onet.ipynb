```python
import glob
import pandas as pd

# Get all CSV files in a target directory
DATA_DIR = "../../data/formatted_onet/"
OUTPUT_FILE = DATA_DIR + "all_merged_output.csv"
files = glob.glob(DATA_DIR + '*.csv')

# Read all CSVs into a list of DataFrames
df_list = [pd.read_csv(file) for file in files if file != OUTPUT_FILE]
df_list = [df.drop(columns="title") for df in df_list]

# Merge DataFrames together based on the ONET_SOC_CODE
merged_df = df_list[0]
for i in range(1, len(df_list)):
    left = merged_df
    right = df_list[i]
    merged_df = pd.merge(left, right, on="ONET_SOC_CODE", how="outer")

merged_df.to_csv(OUTPUT_FILE, index=False)
```
