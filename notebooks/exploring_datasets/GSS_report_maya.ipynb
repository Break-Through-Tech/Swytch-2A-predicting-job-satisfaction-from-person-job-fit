Next is Task #2: Identify the variables of interest for job fit.

The main idea is: compare what people value in a job with how satisfied they are with their current job.

```python
# Variables we think could be relevant to job satisfaction
candidate_vars = [
    "SATJOB",
    "JOBSAT",
    "JOBMEANS",
    "JOBKEEP",
    "JOBRISE",
    "JOBHONOR",
    "JOBOFF",
    "JOBINTER",
    "JOBINDEP",
    "JOBRESP",
    "JOBPEOP",
    "JOBHELP",
    "JOBSOC",
    "JOBACCMP",
    "JOBSAFE",
    "IMPPROMO",
    "ILIKEJOB",
    "TIMEPDWK",
    "WRKEARN",
    "WRKENJOY",
    "WRKSTAT",
    "OCC10",
    "AGE",
    "EDUC",
    "SEX",
    "RACE",
    "INCOME",
    "MARITAL",
    "YEAR"
]

# Keep only variables that actually exist in the dataset
candidate_vars = [v for v in candidate_vars if v in df.columns]

coverage = pd.DataFrame({
    "Valid": df[candidate_vars].notna().sum(),
    "Missing": df[candidate_vars].isna().sum(),
    "Percent Valid": df[candidate_vars].notna().mean() * 100
})

coverage = coverage.sort_values("Percent Valid", ascending=False)

print(coverage)
```

Task #3: Missing values and outliers after selecting your variables.

```python
model_vars = [
    "SATJOB",
    "AGE",
    "SEX",
    "RACE",
    "EDUC",
    "INCOME",
    "MARITAL",
    "WRKSTAT",
    "OCC10",
    "YEAR"
]
df["model_complete"] = df[model_vars].notna().all(axis=1)

print("Total rows:", len(df))
print("Usable rows:", df["model_complete"].sum())
print("Percent usable:", df["model_complete"].mean() * 100)
```

For categorical variables, `MARITAL`, `WRKSTAT`, and `OCC10` showed noticeable differences in job satisfaction across groups. `RACE` showed some differences, while `SEX` had very similar job-satisfaction distributions across its categories. Based on this exploratory analysis, `AGE`, `INCOME`, `MARITAL`, `WRKSTAT`, and `OCC10` appear to be the most promising features for further analysis of job satisfaction. These relationships are descriptive associations and do not imply that these variables cause differences in job satisfaction.
