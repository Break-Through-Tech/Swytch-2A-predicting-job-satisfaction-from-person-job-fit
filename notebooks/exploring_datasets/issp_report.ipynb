```python
is_united_states = df["country"] == US_CODE
print(f"[INFO] Number of responses from U.S. market only: {df[is_united_states].shape[0]}")
```

```python
# Filter DataFrame to only include rows with a job satisfaction included and in U.S.
df = df[df["job_satisfaction"].notna() & (df["country"] == US_CODE)]
print(f"[INFO] Resulting number of useable survey info (U.S. only and has job satisfaction): {df.shape[0]}")
```

```python
no_onet = df["ONET_SOC_CODE"].isna()
has_isco = df["isco08"].notna()

to_resolve = df[no_onet & has_isco]
print(f"[INFO] Number of survey responses that need O*NET code: {to_resolve.shape[0]}")
```

```python
print(f"[INFO] Number of unique ISCO-2008 codes that need mapping: {to_resolve["isco08"].nunique()}")
print(f"[INFO] Listed ISCO-2008 codes that need mapping: \n{"\n".join(str(int(i)) for i in to_resolve["isco08"].unique())}")
```

# Observation of missing codes
It looks like most of the ISCO-2008 codes that do not have a mapping are "general" codes that have not been mapped down into a specific occupation. This is likely due to the fact that it refers to a category, such as "Manager" and could not be mapped to an SOC-2010 occupation code. 

## Potential ideas to handle this
1. Since there are only 36 codes, it is not unreasonable to map this manually. 
2. Or, as our challenge advisor suggested, we could also choose a specific occupation or occupations to focus on. 
3. Finally, we could potentially map one ISCO-2008 code to multiple O\*NET-SOC 2019 code, but this will add more overhead, as we will need to decide/know what our O\*NET occupation profile dataset will look like, and have to balance that fact that a lot of our occupation data is specific to the O*NET code, so occupations under one group may disagree.
